```python
def analyze_students(marks_dict):
    print("--- STUDENT MARKS REPORT ---")

    # 1. Total aur Average (Mean) nikalna
    total_marks = sum(marks_dict.values())
    total_students = len(marks_dict)
    average_marks = total_marks / total_students
    print(f"Total Students: {total_students}")
    print(f"Average Marks of Class: {average_marks:.2f}")

    # 2. Highest aur Lowest (Range ke liye)
    highest_student = max(marks_dict, key=marks_dict.get)
    lowest_student = min(marks_dict, key=marks_dict.get)

    marks_range = marks_dict[highest_student] - marks_dict[lowest_student]

    print(f"Topper: {highest_student} ({marks_dict[highest_student]} marks)")
    print(f"Lowest: {lowest_student} ({marks_dict[lowest_student]} marks)")
    print(f"Marks Range (Max - Min): {marks_range}")

    # 3. Pass/Fail Classification (Conditionals)
    print("\nStatus:")
    for student, marks in marks_dict.items():
        status = "Pass" if marks >= 40 else "Fail"
        print(f"- {student}: {marks} [{status}]")

# Sample Data (Dictionary)
student_data = {
    "chini": 85,
    "anas": 38,
    "honey": 92,
    "Pooja": 74,}
analyze_students(student_data)
```
